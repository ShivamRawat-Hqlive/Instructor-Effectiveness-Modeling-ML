## Step 2: Define Instructor Effectiveness

There is no single ground-truth label for "effectiveness" in this dataset, so we define a **composite score** built from the three families of signal the data actually offers:

| Pillar | Metrics used | Why |
|---|---|---|
| **Outcomes (50%)** | completion_rate, 1 − dropout_rate, avg_score_improvement, avg_quiz_score | The most direct evidence that learners actually *learned something and finished the course* — the closest thing to a "did the instructor achieve the goal" signal |
| **Engagement (25%)** | avg_watch_time, assignment_submission_rate, forum_activity_rate | Engagement is a leading indicator of outcomes and captures how well the instructor holds learner attention/participation, independent of final scores |
| **Feedback (25%)** | avg_feedback_score, feedback_response_rate | Learner-perceived quality — softer, more subjective, and (per the EDA) prone to positivity bias, so it's weighted meaningfully but not dominantly |

**Why Outcomes gets the most weight:** a core promise to learners is that they'll finish and improve — engagement and satisfaction are means to that end, not the end itself. This is a judgment call, not a fact, and is easy to revisit (see limitations in Step 6).

**Normalization:** each raw metric is min–max scaled to [0, 1] across all batches *before* combining, so that metrics on very different natural scales (e.g. quiz score 0–100 vs. rates 0–1) contribute comparably, and `dropout_rate` is inverted (`1 - dropout_rate`) so that higher is always better across every component.

$$\text{BatchScore} = 0.50 \times \overline{\text{Outcomes}} + 0.25 \times \overline{\text{Engagement}} + 0.25 \times \overline{\text{Feedback}}$$

```python
def minmax(s: pd.Series) -> pd.Series:
    return (s - s.min()) / (s.max() - s.min())

work = df.copy()

# Normalize each raw metric to [0, 1]; invert dropout so higher = better everywhere
work["n_completion"]  = minmax(work["completion_rate"])
work["n_dropout_inv"] = minmax(1 - work["dropout_rate"])
work["n_score_impr"]  = minmax(work["avg_score_improvement"])
work["n_quiz"]        = minmax(work["avg_quiz_score"])
work["n_watch"]       = minmax(work["avg_watch_time"])
work["n_assign"]      = minmax(work["assignment_submission_rate"])
work["n_forum"]       = minmax(work["forum_activity_rate"])
work["n_feedback"]    = minmax(work["avg_feedback_score"])
work["n_fb_resp"]     = minmax(work["feedback_response_rate"])

outcome_pillar    = work[["n_completion", "n_dropout_inv", "n_score_impr", "n_quiz"]].mean(axis=1)
engagement_pillar = work[["n_watch", "n_assign", "n_forum"]].mean(axis=1)
feedback_pillar   = work[["n_feedback", "n_fb_resp"]].mean(axis=1)

work["batch_effectiveness_score"] = (
    0.50 * outcome_pillar + 0.25 * engagement_pillar + 0.25 * feedback_pillar
)

plt.figure(figsize=(7,4))
sns.histplot(work["batch_effectiveness_score"], bins=25, kde=True, color="#8172B2")
plt.title("Distribution of Batch-Level Effectiveness Score")
plt.show()

work["batch_effectiveness_score"].describe()
```

Tiering happens **after** aggregation to the instructor level (Step 3), since effectiveness is fundamentally a statement about an *instructor*, not a single batch.

## Step 3: Aggregate Batch Data to Instructor Level

**Aggregation choices:**
- **Mean** for every rate/score metric — the natural summary of "typical performance" across an instructor's batches, and robust to the moderate batch counts here (7–31).
- **Count of batches (`n_batches`)** kept as its own feature — this lets the model learn to weight instructors with more evidence differently, instead of silently pretending a 7-batch and a 31-batch instructor are equally reliable.
- **Standard deviation of the batch score (`score_std`)** kept as a *consistency* feature — an instructor who is reliably "good" across batches is arguably more effective (and more trustworthy to evaluate) than one who is wildly inconsistent even if their mean is the same. Instructors with only one batch get `score_std = 0` (no variance is observable) rather than `NaN`, so they aren't dropped.

**Handling few vs. many batches:** we do *not* discard low-batch-count instructors — with only 120 instructors total, dropping any would waste scarce data and could itself be an equity issue (e.g. new instructors would always be excluded). Instead, `n_batches` and `score_std` are passed into the model explicitly, so it can learn to be appropriately less confident about (or discount) sparse-data instructors rather than that reliability information being silently lost.

```python
instructor_df = df.groupby("instructor_id").agg(
    n_batches=("batch_id", "count"),
    completion_rate=("completion_rate", "mean"),
    dropout_rate=("dropout_rate", "mean"),
    avg_score_improvement=("avg_score_improvement", "mean"),
    avg_quiz_score=("avg_quiz_score", "mean"),
    avg_watch_time=("avg_watch_time", "mean"),
    assignment_submission_rate=("assignment_submission_rate", "mean"),
    forum_activity_rate=("forum_activity_rate", "mean"),
    avg_feedback_score=("avg_feedback_score", "mean"),
    feedback_response_rate=("feedback_response_rate", "mean"),
).reset_index()

# Instructor-level score = mean of that instructor's batch-level scores;
# score_std captures consistency across batches (0 if only one batch)
score_stats = work.groupby("instructor_id")["batch_effectiveness_score"].agg(
    instructor_score="mean", score_std="std"
).reset_index()
score_stats["score_std"] = score_stats["score_std"].fillna(0)

instructor_df = instructor_df.merge(score_stats, on="instructor_id")
print(instructor_df.shape)
instructor_df.head()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(data=instructor_df, x="n_batches", y="instructor_score", hue="score_std",
                 palette="viridis", size="score_std", sizes=(20,120))
plt.title("Instructor Score vs. Number of Batches Taught")
plt.xlabel("Number of batches")
plt.ylabel("Instructor effectiveness score")
plt.show()
```

**Observation:** score doesn't trend up or down with `n_batches` — instructors who teach more batches aren't systematically scored higher or lower, which is reassuring (no obvious volume bias baked into the mean-based aggregation).

### Effectiveness Tiers

We convert the continuous `instructor_score` into **Low / Medium / High** tiers using **tertile (33rd/67th percentile) cut-offs**, chosen over fixed absolute thresholds because:
- Effectiveness is inherently relative in this dataset — there's no external absolute benchmark (e.g. "80% completion = good") provided, so relative ranking against peers is the most defensible interpretation.
- Tertiles guarantee roughly balanced classes, which keeps the downstream classification task meaningful (no tier with 2 members).

This is a **design choice, not the only valid one** — K-means clustering on the score, or business-defined absolute thresholds if the company has them, are equally reasonable alternatives.

```python
q_low, q_high = instructor_df["instructor_score"].quantile([1/3, 2/3])
print(f"Tertile cut-offs: Low <= {q_low:.3f} < Medium <= {q_high:.3f} < High")

def tier_from_score(s):
    if s <= q_low:
        return "Low"
    elif s <= q_high:
        return "Medium"
    else:
        return "High"

instructor_df["effectiveness_tier"] = instructor_df["instructor_score"].apply(tier_from_score)
tier_order = ["Low", "Medium", "High"]

print(instructor_df["effectiveness_tier"].value_counts()[tier_order])

plt.figure(figsize=(6,4))
sns.countplot(data=instructor_df, x="effectiveness_tier", order=tier_order, palette="Set2")
plt.title("Instructor Effectiveness Tier Distribution")
plt.show()
```

## Step 4: Build a Machine Learning Model

**Target:** `effectiveness_tier` (Low / Medium / High), at the **instructor level** (n = 120).

**Model choice: Random Forest Classifier.** Reasoning:
- Small tabular dataset with mixed-scale numeric features — tree ensembles handle this well without heavy preprocessing.
- Naturally handles non-linear thresholds/interactions (e.g. "high engagement only helps if dropout is also low").
- Gives interpretable feature importances for Step 6, which matters more here than squeezing out maximum accuracy per the brief.
- Robust to the modest multicollinearity noted in the EDA (completion vs. dropout), unlike e.g. plain logistic regression.

**Important caveat (flagged now, revisited in Step 6/limitations):** the label itself (`effectiveness_tier`) is a **deterministic function of the same features** the model is trained on. This makes the classification task somewhat circular — a good model is mostly "re-deriving" our own formula rather than discovering an independent, externally-validated notion of effectiveness. We proceed as instructed (this is the objective set by the task), but this is an important limitation to be transparent about, and it's exactly why the mandatory questions about confounds and real-world failure modes matter here.

```python
feature_cols = [
    "completion_rate", "dropout_rate", "avg_score_improvement", "avg_quiz_score",
    "avg_watch_time", "assignment_submission_rate", "forum_activity_rate",
    "avg_feedback_score", "feedback_response_rate", "n_batches", "score_std"
]

X = instructor_df[feature_cols]
y = instructor_df["effectiveness_tier"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
print("Train size:", X_train.shape, " Test size:", X_test.shape)
print(y_train.value_counts())
print(y_test.value_counts())
```

```python
# Scaling isn't required for tree models but is applied for consistency /
# so this pipeline works if the model is swapped for a scale-sensitive one later.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

clf = RandomForestClassifier(
    n_estimators=300,
    max_depth=5,
    class_weight="balanced",
    random_state=RANDOM_STATE
)
clf.fit(X_train_scaled, y_train)

y_pred = clf.predict(X_test_scaled)
```

## Step 5: Evaluate the Model

Given only 120 instructors (24 in the test set), we look beyond raw accuracy:
- **Macro-F1** — since tiers were constructed to be balanced (40/40/40), macro-F1 and accuracy should broadly agree, but macro-F1 still protects against any one class being systematically confused.
- **Per-class precision/recall** — specifically, false "High" predictions (precision) matter more than false negatives if this were ever used for rewards, while missed "Low" instructors (recall) matter more if it's used to trigger support/coaching.
- **Confusion matrix** — to see *which* tiers get confused (adjacent-tier mistakes like Medium↔High are far less concerning than Low↔High mistakes).

```python
acc = accuracy_score(y_test, y_pred)
macro_f1 = f1_score(y_test, y_pred, average="macro")
print(f"Accuracy: {acc:.3f}")
print(f"Macro F1: {macro_f1:.3f}")
print()
print(classification_report(y_test, y_pred, labels=tier_order))
```

```python
fig, ax = plt.subplots(figsize=(5.5,5))
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred, labels=tier_order, cmap="Blues", ax=ax, colorbar=False
)
plt.title("Confusion Matrix — Effectiveness Tier")
plt.show()
```

**Class imbalance:** by construction (tertile split), the three tiers are almost perfectly balanced (40/40/40 at the instructor level, and correspondingly ~stratified in the train/test split), so class imbalance is **not a real concern here** — this is a deliberate consequence of choosing tertile-based tiering in Step 3. If tiers were instead defined by fixed absolute thresholds, imbalance could easily appear (e.g. very few "Low" instructors), in which case `class_weight="balanced"` (already used above), macro-F1 over accuracy, and/or resampling would become important.

**Precision vs. recall trade-off:** any confusion involving the "Low" tier deserves the most scrutiny, since this is the tier most likely to trigger a real-world consequence (coaching, review, or performance action) for an instructor — a false "Low" prediction (low precision on that class) risks unfairly flagging an effective instructor, while a missed "Low" (low recall) means a struggling instructor goes unsupported. Given the stakes, we'd lean toward prioritizing precision on "Low" if this were deployed, i.e. requiring high confidence before flagging someone as underperforming.

## Step 6: Interpret the Results

```python
importances = pd.Series(clf.feature_importances_, index=feature_cols).sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=importances.values, y=importances.index, palette="mako")
plt.title("Feature Importance — Random Forest")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

importances
```

**In plain terms (non-technical summary):**

The model relies most heavily on **completion rate**, **dropout rate**, and **feedback response rate** to judge an instructor's effectiveness tier — in other words, *whether learners actually finish the course, don't drop out, and bother to give feedback at all* are the strongest signals. Score improvement, quiz performance, and average feedback rating matter too, but less. Interestingly, raw engagement metrics like watch time, assignment submission, and forum activity contribute the least — suggesting that in this dataset, whether learners *stick around and complete* carries more information than how actively they engage along the way.

**How this could be used in a Tech product:** this kind of scoring could power an internal instructor dashboard that flags batches trending toward high dropout early (before the batch even ends), or surfaces instructors in the "Low" tier for proactive support (e.g. content review, coaching, pairing with a high-tier instructor) rather than punitive action — framed as a support tool, not a performance-management verdict (see Section 5 answers below for why).

## Section 5: Mandatory Analysis Questions

**Q1. Which features most influenced instructor effectiveness, and why?**
`completion_rate`, `dropout_rate`, and `feedback_response_rate` dominate the feature importances. This tracks with how the score itself was constructed (Outcomes = 50% weight), but it's worth noting the model *discovered* that outcome-related features matter more than engagement features even *within* the 50/25/25 split — i.e., the outcome pillar's individual metrics are more decisive/less noisy than the individual engagement or feedback metrics, which is plausible since completion/dropout are hard, unambiguous events while engagement metrics (forum activity especially) are noisier and more optional-behavior-dependent.

**Q2. Which variables could be misleading or confounded?**
- `avg_feedback_score` and `feedback_response_rate` can reflect course *content* difficulty, topic popularity, or grading leniency as much as instructor skill — a hard but valuable course may get lower ratings than an easy but shallow one.
- `forum_activity_rate` depends heavily on whether a course's format even encourages forum use (a hands-on coding course vs. a discussion-heavy humanities course), so it may say more about *course design* than about the *instructor*.
- Instructors are also implicitly confounded with **course** and **batch cohort** (e.g. time of year, learner motivation level, marketing channel that brought the learners in) — none of which are captured in this dataset, yet all of which affect completion and dropout independent of instructor quality.
- `n_batches` could correlate with tenure/seniority rather than skill — more experienced instructors might get assigned differently (e.g. harder or easier courses), which is itself a form of selection bias.

**Q3. How could this model fail in real-world usage?**
- **Circularity (flagged in Step 4):** since the label is derived from the same features used to predict it, the model mostly learns to reproduce our formula rather than an independently validated notion of "effectiveness" — it offers little that a simple weighted score alone wouldn't.
- **Small sample size:** 120 instructors (96 train / 24 test) is a small dataset for a tree ensemble; results can be unstable to the train/test split and shouldn't be treated as a robust, generalizable estimate.
- **No course/topic control:** an instructor consistently assigned difficult, low-completion courses could be misclassified as "Low" effectiveness through no fault of their own.
- **Feedback response bias:** learners who leave feedback are self-selected (often more/less satisfied than average), so `feedback_response_rate` and `avg_feedback_score` may not represent the full cohort.
- **Static snapshot:** the model has no sense of trend (an instructor improving over time looks identical to one declining, if their mean is the same).

**Q4. What additional data would you want to improve this analysis?**
- Course difficulty/level and subject area, to control for course-driven variation in completion/dropout.
- Learner-level data (prior experience, motivation, time zone/schedule fit) to separate learner-driven attrition from instructor-driven attrition.
- Cohort context: batch size, marketing source, price/free tier, start season.
- Time-series data across an instructor's tenure, to detect trends rather than a static average.
- Independent effectiveness signals not derived from the same batch metrics — e.g. peer/manager reviews, learner re-enrollment or referral rate, post-course outcomes (job placement, skill assessments weeks later).

**Q5. Should this model be used for instructor performance evaluation? Why or why not?**
Not on its own, and not for high-stakes decisions (pay, termination) in its current form. It's a reasonable **triage / support tool** — e.g. surfacing instructors or batches worth a closer human look — but given the circularity of the label, the lack of course/cohort controls, and the small sample size, treating its tier output as a verdict on an instructor risks penalizing people for factors outside their control (course assignment, learner mix) rather than their actual teaching. If used at all, it should be paired with qualitative review, applied at the course level (comparing instructors *within* the same course) rather than across different courses, and framed to instructors transparently rather than used silently.

## Summary

- Built a transparent, justified **Instructor Effectiveness Score** (50% outcomes / 25% engagement / 25% feedback, all min-max normalized) at the batch level, then aggregated to the instructor level using means (plus `n_batches` and `score_std` to preserve reliability/consistency information).
- Converted the score into **balanced Low/Medium/High tiers** via tertiles.
- Trained a **Random Forest** classifier reaching **~92% test accuracy / ~0.92 macro-F1** on a small (n=120) instructor-level dataset — a reasonable but not over-claimed result given the sample size and the label's circularity.
- Outcome metrics (completion, dropout, feedback response) drove the model far more than engagement metrics.
- Flagged the model's real limitations (circular target, missing course/cohort controls, self-selection in feedback) rather than overselling accuracy, per the brief's emphasis on reasoning over raw performance.
